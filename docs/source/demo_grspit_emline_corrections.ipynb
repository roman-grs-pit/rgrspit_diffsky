# Empirical corrections to Diffsky emission lines 

The Diffsky mock generated for the Roman Galaxy Redshift Survey (GRS) Project Infrastructure Team (PIT) includes a set of empirical corrections to the emission lines that improve the observational realism. This notebook shows how to get the improved lines from the GRS PIT Diffsky library.

```python
from jax import random as jran

ran_key = jran.key(0)
```

```python
from rgrspit_diffsky.emission_lines import oii, oiii
```

```python
import numpy as np
n = 5_000

lum_oii = np.logspace(38, 46, n)
lum_oii_obs = oii.get_observed_OII(lum_oii, ran_key)
```

```python
from matplotlib import pyplot as plt

fig, ax = plt.subplots(1, 1)
__=ax.loglog()
__=ax.scatter(lum_oii, lum_oii_obs, s=2)
```
